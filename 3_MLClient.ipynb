###  Creating the Client

```python
import requests
import json

data = """{
    "row.names": 1,
    "sbp": 160,
    "tobacco": 4.8,
    "ldl": 8.63,
    "adiposity": 36.21,
    "famhist": "Present",
    "typea": 50,
    "obesity": 34.72,
    "alcohol": 28.8,
    "age": 61
}"""

response = requests.post(
    "http://127.0.0.1:5002/predict",
    json=json.loads(data)
)

print(response.status_code)
print(response.text)
```

```python
import requests

url = "https://mlops-2-6l4p.onrender.com/predict"

data = """{
    "row.names": 1,
    "sbp": 160,
    "tobacco": 4.8,
    "ldl": 8.63,
    "adiposity": 36.21,
    "famhist": "Present",
    "typea": 50,
    "obesity": 34.72,
    "alcohol": 28.8,
    "age": 61
}"""
response = requests.post(
    url,
    json=data
)

print("Status Code:", response.status_code)
print("Response:", response.json())
```
